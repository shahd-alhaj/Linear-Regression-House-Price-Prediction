```python
import numpy as np
import matplotlib.pyplot as plt

print("My first Machine Learning Lab!")
```

```python

x_train = np.array([1.0, 2.0])


y_train = np.array([300.0, 500.0])

print("House Sizes:", x_train)
print("House Prices:", y_train)
```

```python

plt.scatter(x_train, y_train, marker='x', c='r')


plt.title("Housing Prices")
plt.xlabel("House Size (1000 sqft)")
plt.ylabel("Price (1000s of dollars)")


plt.show()
```

```python

w = 200
b = 100


y_pred = w * x_train + b


plt.scatter(x_train, y_train, marker='x', c='r',
            label='Actual Values')

plt.plot(x_train, y_pred, c='b',
         label='Our Prediction')


plt.title("Housing Prices - Linear Regression")
plt.xlabel("House Size (1000 sqft)")
plt.ylabel("Price (1000s of dollars)")
plt.legend()
plt.show()
```

```python

house_size = 1.2

predicted_price = w * house_size + b

print("Linear Regression - House Price Prediction")
print(f"House Size: {house_size * 1000:.0f} sqft")
print(f"Predicted Price: ${predicted_price * 1000:,.0f}")
```
